```python
import os
import cv2
from ultralytics import YOLO


# Model used
MODEL_VERSIONS = {
    "v8":  "yolov8n.pt",
    "v11": "yolo11n.pt",
    "v26": "yolo26n.pt",
}


def save_image(filepath, img):
    return cv2.imwrite(filepath, img)


def detect_and_view(image_path, model_path="yolo26n.pt", conf_threshold=0.3, output_path=None):

    # Validate input file
    for path in (image_path, model_path):
        if not os.path.exists(path):
            print(f"Error: '{path}' file not found!")
            return

    img = cv2.imread(image_path)
    if img is None:
        print(f"Error: can't read image '{image_path}'.")
        return

    # Load model
    try:
        model = YOLO(model_path)
    except Exception as e:
        print(f"Error: gagal load model '{model_path}' -> {e}")
        return

    result = model.predict(
        source=img,
        conf=conf_threshold,
        verbose=False,
    )[0]
    boxes = result.boxes

    if len(boxes) == 0:
        print(f"No objects detected (conf >= {conf_threshold:.0%}).")
        return

    print(f"Detected {len(boxes)} object(s) using '{os.path.basename(model_path)}':\n")

    for i, box in enumerate(boxes, 1):
        class_name = model.names[int(box.cls[0])]
        conf = float(box.conf[0])
        xmin, ymin, xmax, ymax = map(int, box.xyxy[0])

        print(f"  {i}. {class_name} ({conf:.2%}) -> [x1:{xmin}, y1:{ymin}, x2:{xmax}, y2:{ymax}]")

    annotated_frame = result.plot()

    if output_path:
        save_image(output_path, annotated_frame)

    cv2.imshow("Output", annotated_frame)
    cv2.waitKey(0)
    cv2.destroyAllWindows()


if (__name__ == "__main__"):
    BASE_DIR = (
        r"D:/Amr/Portfolio/All time project/ready to publish/DL/project 6"
    )

    for version, model_filename in MODEL_VERSIONS.items():
        print(f"\n{version} ({model_filename})")
        detect_and_view(
            image_path=os.path.join(BASE_DIR, "input.png"),
            model_path=os.path.join(BASE_DIR, model_filename),
            conf_threshold=0.3,
            output_path=os.path.join(BASE_DIR, f"output_{version}.png"),
        )
```

```python
# Object detection from video

def run_yolo_video(source=0, model_size="yolov8n.pt", conf_threshold=0.4):

    # Load model 
    model = YOLO(model_size)

    # Stream and show result
    model.predict(
        source=source,
        conf=conf_threshold,
        show=True, 
        save=True,  
        verbose=False
    )

if __name__ == "__main__":
    video_path = r"D:/Amr/Portfolio/All time project/ready to publish/DL/project 6/Video_input.mp4"
    run_yolo_video(source=video_path, model_size="yolov8n.pt", conf_threshold=0.4)
```
